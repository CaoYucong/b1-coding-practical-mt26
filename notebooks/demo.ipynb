# Simulate closed-loop
After implementing your control functionality, you can simulate the closed-loop with code that looks something like this...

```python
# Import the submarine model, mission, closed-loop simulation, and PD controller.
from uuv_mission.dynamic import Submarine, ClosedLoop, Mission
from uuv_mission.control import PDController

# Create the submarine and controller, then connect them in closed loop.
submarine = Submarine()
controller = PDController()
closed_loop = ClosedLoop(submarine, controller)

# Load the mission data.
mission = Mission.from_csv("../data/mission.csv")

# Run the closed-loop simulation with random disturbances and plot the trajectory.
trajectory = closed_loop.simulate_with_random_disturbances(mission)
trajectory.plot_completed_mission(mission)
```
